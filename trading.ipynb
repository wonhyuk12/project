```python
import random

# 1. 환경 설정 (4x4 미로)
# 맵 생성
grid = [
    ['S', '.', '.', '.'],
    ['.', 'H', '.', 'H'],
    ['.', '.', '.', 'H'],
    ['H', '.', '.', 'G']
]
for i in range(len(grid)):
    print(grid[i])
# 행동 정의: 0: 위, 1: 아래, 2: 왼쪽, 3: 오른쪽
actions = [(-1, 0), (1, 0), (0, -1), (0, 1)]

# 2. Q-Table 초기화 (모든 상태-행동 조합을 0.0으로 초기화)
q_table = {}
for r in range(4):
    for c in range(4):
        q_table[(r, c)] = [0.0, 0.0, 0.0, 0.0]

# 하이퍼파라미터
alpha = 0.1     # 학습률 (Learning Rate): 새 정보를 얼마나 반영할가
gamma = 0.9     # 할인율 (Discount Factor): 미래의 보상을 얼마나 중요하게 볼가
epsilon = 0.1   # 탐험률 (Exploration Rate): 20% 확률로 랜덤 행동

# 3. 학습 과정 (1000번의 에피소드 반복)
for episode in range(100):
    state = (0, 0) # 시작 위치
    
    while True:
        # 종료 조건 (출구나 함정에 도달한 경우)
        if grid[state[0]][state[1]] in ['G', 'H']:
            break
            
        # e-Greedy 정책: 탐험(랜덤) vs 활용(표에서 가장 높은 점수 선택)
        if random.random() < epsilon:
            action_idx = random.randint(0, 3) # 랜덤 탐험
        else:  
            action_idx = q_table[state].index(max(q_table[state])) # 현재 최고 점수 행동 선택
            
        # 다음 상태 계산
        dr, dc = actions[action_idx]
        next_r = max(0, min(3, state[0] + dr))
        next_c = max(0, min(3, state[1] + dc))
        next_state = (next_r, next_c)
        
        # 보상 계산
        cell = grid[next_r][next_c]
        if cell == 'G':
            reward = 100
        elif cell == 'H':
            reward = -10
        else:
            reward = -1  # 빠른 길을 찾도록 걸음마다 약간의 감점
            
        # --- [Q-Learning의 핵심: Q-Table 업데이트] ---
        old_value = q_table[state][action_idx]
        next_max_q = max(q_table[next_state])
        
        # Q-Learning 공식 적용
        new_value = old_value + alpha * (reward + gamma * next_max_q - old_value)
        q_table[state][action_idx] = new_value
        
        state = next_state # 상태 이동

print("=== 학습 완료! Q-Table 결과 ===")
for r in range(4):
    for c in range(4):
        best_act = ["위", "아래", "왼쪽", "오른쪽"][q_table[(r, c)].index(max(q_table[(r, c)]))]
        print(f"위치 ({r},{c}) 추천 행동: {best_act} | Q값: {[round(v,1) for v in q_table[(r,c)]]}")
```
